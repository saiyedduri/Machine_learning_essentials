# Random vectors:
each feature/column in a dataframe is called a random vector. X = (X1, ..., Xn)^T

A random vector is just several random variables bundled together into one vector:
X : Ω → ℝⁿ

# Random Matrix:
family of random vectors form a random matrix.
X = (Xij), i=1,...,n; j=1,...,m

The realization of random variables are represented with lower case letters. X(ω) = x

# Distribution of random vector:
probabilities that realizations of the random vector are contained in suitable subsets of ℝⁿ.
P_X(B) = P(X ∈ B),   B ⊆ ℝⁿ, suitable.

If the distribution of the random vector is given then the random vector X is given.

# Cumulative distribution of random vector X = (X1,...,Xn):
The distribution of an n-dimensional random vector X is
F_X : ℝⁿ → [0, 1]:

F_X(x) = P_X((-∞ⁿ, x]) = P(X ≤ x),   x ∈ ℝⁿ.
F_X(x) = P(X1 ≤ x1, ..., Xn ≤ xn)

# Properties of cumulative distribution function of random vector:

(i)    lim_{xi → -∞, i∈{1,...,n}} F_X(x) = 0 ;

(ii)   lim_{xi → ∞, i=1,...,n} F_X(x) = 1 ;

(iii)  for arbitrary a < b it holds

           Δ_{1;a1,b1} ... Δ_{n;an,bn} F_X(x) ≥ 0 ,     (= P_X((a,b]))

       ⇒  F_X is non-decreasing with respect to each variable

(iv)   F_X is continuous from the right with respect to each variable, i.e.,
       from x^(k) ↓ x it follows F_X(x^(k)) ↓ F_X(x) ,  k → ∞

# Properties of probability densities

- for all x ∈ ℝⁿ :   f_X(x) ≥ 0

- ∫_{ℝⁿ} f_X(x) dx = 1

- then it holds for suitable B ⊆ ℝⁿ

        P(X ∈ B) = ∫_B f_X(x) dx

# Marginal distributions

- the marginal distribution function of a subvector of a random vector
  can be calculated from the distribution function of the original
  random vector

- Example: X = (X1, ..., Xn)^T  ⇒  ∀ x1 ∈ ℝ :

        F_X1(x1) = P(X1 ≤ x1) = F_X(x1, ∞, ..., ∞).

- analogously marginal densities can be calculated for subvectors of
  absolutely continuous random vectors

- Example: X = (X1, ..., Xn)^T  ⇒  ∀ x1 ∈ ℝ :

        f_X1(x1) = ∫_{ℝ^(n-1)} f_X(x1, x2, ..., xn) dx2 ... dxn .

- only in special cases one can determine the distribution function or
  density of a random vector only from the marginal distribution
  functions or densities

# Cramér–Wold Theorem:
if you know the 1-D distribution of t^T X for every possible t ∈ ℝⁿ, you can fully reconstruct the joint (n-dimensional) distribution of X.

# Expectation of a random vector

- expectations of random vectors or matrices are defined
  component-wise, if the expectation of each entry (component) exists

- expectation of the random vector X = (X1, ..., Xn)^T :

        EX := (EX1, ..., EXn)^T

- expectation of the random matrix X = (Xij):

        EX := (EXij)

# Covariance matrix of random vector:

- covariance matrix of X = (X1, ..., Xn)^T :

        Var X := E[ [X − EX][X − EX]^T ]

- main diagonal entries: variances Var Xi of the components

- off-diagonal entries (i,j), i≠j: covariance Cov[Xi, Xj]

- properties of Σ = Var X

        Σ = Σ^T                       (Σ is symmetric)

        ∀ x ∈ ℝⁿ :  x^T Σ x ≥ 0        (Σ is positive semidefinite)

# Correlation matrix of random vector:

if 0 < Var Xi < ∞ for all components of X = (X1, ..., Xn)^T

⇒ one can define the correlation matrix of X:

        Corr X := ( Corr[Xi, Xj] )_{i,j=1,...,n}

with

        Corr[Xi, Xj] := Cov[Xi, Xj] / sqrt(Var Xi · Var Xj)

- the correlation matrix of the random vector X = (X1, ..., Xn)^T is
  the covariance matrix of the standardized components of X.

- off-diagonal entry at place (i,j) with i≠j is the correlation
  coefficient between Xi and Xj.

- it holds always

        -1 ≤ Corr[Xi, Xj] ≤ 1

- in the case |Corr[Xi, Xj]| = 1 there is a linear
  relationship between random variables Xi and Xj.

# Cross covariance between two random vectors

- if $X = (X_1,\ldots,X_n)^T$ and $Y = (Y_1,\ldots,Y_m)^T$ have components
  with finite second-order moments, the **cross covariance matrix** is
  the $n \times m$ matrix

$$
\mathrm{Cov}[X,Y] := E\big[ [X-EX][Y-EY]^T \big] .
$$

- entry $(i,j)$ = $\mathrm{Cov}[X_i, Y_j]$

- special case: $\mathrm{Var}\,X = \mathrm{Cov}[X,X] =: \mathrm{Cov}\,X$

- symmetry rule: $\mathrm{Cov}[X,Y] = \mathrm{Cov}[Y,X]^T$

- analogously defines the **cross correlation matrix** $\mathrm{Corr}[X,Y]$

- $X$ and $Y$ are **uncorrelated** if $\mathrm{Cov}[X,Y] = 0_{n\times m}$

# Properties related to linear operations

**Given:** $X, Y$ $n$-dim. r.v., $E\|X\| < \infty$, $E\|Y\| < \infty$.

- $a, b \in \mathbb{R} \ \Rightarrow \ E[aX + bY] = a\,EX + b\,EY$ ;

- $d \in M_{m \times n}, c \in \mathbb{R}^m \ \Rightarrow \ E[dX + c] = d\,EX + c$ .

**Given:** $X$ $n$-dim. r.v., $E\|X\|^2 < \infty$.

- $\mathrm{Var}\,X = E[XX^T] - (EX)(EX)^T$ ;

- $a \in \mathbb{R}^n \ \Rightarrow \ \mathrm{Var}[a^TX] = a^T \mathrm{Var}[X]\, a$ ;

- $a \in M_{m \times n}, b \in \mathbb{R}^m \ \Rightarrow \ \mathrm{Var}[aX + b] = a\,\mathrm{Var}[X]\,a^T$ .

**Given:** $X^{(1)}, X^{(2)}$ $n$-dim. r.v., $Y$ $m$-dim. r.v.,
$E\|X^{(i)}\|^2 < \infty$, $i=1,2$, $E\|Y\|^2 < \infty$

$$
\Rightarrow \quad \mathrm{Cov}\big[X^{(1)}+X^{(2)}, Y\big] = \mathrm{Cov}\big[X^{(1)}, Y\big] + \mathrm{Cov}\big[X^{(2)}, Y\big]
$$

# Properties related to linear operations II

**Given:** $X, Y$ $n$-dim. r.v., $E\|X\|^2 < \infty$, $E\|Y\|^2 < \infty$

$$
\Rightarrow \quad \mathrm{Var}[X+Y] = \mathrm{Var}\,X + \mathrm{Cov}[X,Y] + \mathrm{Cov}[Y,X] + \mathrm{Var}\,Y .
$$

**Given:** $X$ $n_1$-dim. r.v., $Y$ $n_2$-dim. r.v.,
$E\|X\|^2 < \infty$, $E\|Y\|^2 < \infty$,
$a \in M_{m_1 \times n_1}$, $b \in M_{m_2 \times n_2}$

$$
\Rightarrow \quad \mathrm{Cov}[aX, bY] = a\, \mathrm{Cov}[X,Y]\, b^T .
$$

# Independence of random vectors

**Given:** $X$ $n$-dim. r.v., $Y$ $m$-dim. r.v., $Z = (X^T, Y^T)^T$

$\Rightarrow$ $X$ and $Y$ are (stochastically) **independent**, if one of the
following equivalent conditions holds
(condition (iii) only if $Z$ is (absolutely) continuous)

**(i)** $\forall B_1 \subseteq \mathbb{R}^n, \forall B_2 \subseteq \mathbb{R}^m$, suitable:
$$
P(\{X \in B_1\} \cap \{Y \in B_2\}) = P(X \in B_1) \cdot P(Y \in B_2) .
$$

**(ii)** $\forall x \in \mathbb{R}^n, \forall y \in \mathbb{R}^m$:
$$
F_Z(x,y) = F_X(x) \cdot F_Y(y) .
$$

**(iii)** $\forall x \in \mathbb{R}^n, \forall y \in \mathbb{R}^m$:
$$
f_Z(x,y) = f_X(x) \cdot f_Y(y) .
$$

- if second-order moments of all components exist, independent random
  vectors are also **uncorrelated**

# Multidimensional normal distribution

**Def. 2.2.1** An $m$-dimensional random vector $X = (X_1,\ldots,X_m)^T$
has an $m$-dimensional **standard normal distribution**, iff

$$
X_i \sim N(0,1) , \quad i=1,\ldots,m , \ \text{i.i.d.}
$$

**Notation:** $X \sim N_m(0_m, I_m)$ or $X \sim N(0_m, I_m)$

**Theorem 2.2.2** Given: random vector $X \sim N_m(0_m, I_m)$

$\Rightarrow$ $X$ is a continuous random vector with density function

$$
f_X(x) = (2\pi)^{-m/2} e^{-\frac{1}{2}x^Tx} = (2\pi)^{-m/2} e^{-\frac{1}{2}\|x\|^2} , \quad x \in \mathbb{R}^m .
$$

Furthermore it holds $EX = 0_m$ and $\mathrm{Var}\,X = I_m$ .

# General normally distributed random vectors

**Def. 2.2.3** A $p$-dimensional random vector $X = (X_1,\ldots,X_p)^T$
has a $p$-dimensional **normal distribution**, if for an $m$-dimensional
standard normally distributed random vector $Z$, a vector $\mu \in \mathbb{R}^p$
and a $p \times m$ matrix $a$ it holds

$$
X = \mu + aZ .
$$

**Theorem 2.2.4** For the random vector $X$ from Def. 2.2.3 it holds

$$
EX = \mu \qquad \text{and} \qquad \mathrm{Var}\,X = a\,a^T =: \Sigma .
$$

**Notation:** $X \sim N_p(\mu, \Sigma)$ or $X \sim N(\mu, \Sigma)$ .
Also the name **Gaussian random vector** is used.

**Remark:** The $p \times p$ matrix $a\,a^T = \Sigma$ is (e.g. as a
covariance matrix) symmetric and positive semidefinite.

**Theorem 2.2.6** A $p$-dimensional random vector $X = (X_1,\ldots,X_p)^T$
is normally distributed iff for each vector $t \in \mathbb{R}^p$ the
real-valued random variable $t^TX$ has a normal distribution or is a
constant.

**Theorem 2.2.7** Let $X = (X_1,\ldots,X_p)^T \sim N_p(\mu, \Sigma)$ .

**(i)** For $b \in \mathbb{R}^d$, $a \in M_{d \times p}$ it holds

$$
Y := b + aX \ \sim \ N_d(b + a\mu, \ a\Sigma a^T) .
$$

**(ii)** Each subvector and each component of $X$ is a normally
distributed random vector or variable respectively.

**Remark** There exist random vectors which are not normally
distributed, but every component is a normally distributed random
variable.

**Theorem 2.2.7** Let $X = (X_1,\ldots,X_p)^T \sim N_p(\mu, \Sigma)$ .

**(i)** For $b \in \mathbb{R}^d$, $a \in M_{d \times p}$ it holds

$$
Y := b + aX \ \sim \ N_d(b + a\mu, \ a\Sigma a^T) .
$$

**(ii)** Each subvector and each component of $X$ is a normally
distributed random vector or variable respectively.

# Random vectors with a regular normal distribution

**Def. and Theorem 2.2.8** Let $X$ be a normally distributed random
vector according to Definition 2.2.3 with $p = m$, $\mathrm{Var}\,X = a\,a^T =: \Sigma$
such that $\det \Sigma \neq 0$ . Then one says that $X$ has a
**regular normal distribution** $N_p(\mu, \Sigma)$ .

$X$ is a continuous random vector with density function

$$
f_X(x) = \frac{1}{\sqrt{(2\pi)^p \det \Sigma}} \, e^{-\frac{1}{2}(x-\mu)^T \Sigma^{-1} (x-\mu)} , \quad x \in \mathbb{R}^p .
$$

The normalising factor can also be written as $\det(2\pi\Sigma)^{-1/2}$ .

$\Sigma$ is a real symmetric and positive definite $p \times p$ matrix,
so that it holds $\det \Sigma > 0$ .

# Independence of subvectors in case of a normal distribution

**Theorem 2.2.9** Let $X = (X_1^T, X_2^T)^T \sim N_p(\mu, \Sigma)$,
$X_1 = (X_1,\ldots,X_k)^T$, $X_2 = (X_{k+1},\ldots,X_p)^T$ .

Then $X_1$ and $X_2$ are stochastically independent iff it holds

$$
\mathrm{Cov}[X_1, X_2] = 0_{k \times (p-k)} .
